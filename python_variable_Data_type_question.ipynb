```python
# 21. Create a Student Report using variables and data types.

student_name = input("Enter student name: ")
age = int(input("Enter age: "))
student_class = int(input("Enter class: "))
roll_number = int(input("Enter roll number: "))
english_marks = int(input("Enter English marks: "))
maths_marks = int(input("Enter Maths marks: "))
science_marks = int(input("Enter Science marks: "))
height = float(input("Enter height: "))
student_status = input("Is student active? (True/False): ") == "True"

print("\n--- Student Report ---")
print("Student Name:", student_name, type(student_name))
print("Age:", age, type(age))
print("Class:", student_class, type(student_class))
print("Roll Number:", roll_number, type(roll_number))
print("English Marks:", english_marks, type(english_marks))
print("Maths Marks:", maths_marks, type(maths_marks))
print("Science Marks:", science_marks, type(science_marks))
print("Height:", height, type(height))
print("Student Status:", student_status, type(student_status))
```
